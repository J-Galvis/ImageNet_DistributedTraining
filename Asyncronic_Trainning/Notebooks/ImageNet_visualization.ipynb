```python
import json
import sys
import os
from pathlib import Path
from typing import Optional

import altair as alt
import pandas as pd

# Añadir el directorio padre al path para acceder al paquete Utils
notebook_dir = os.path.dirname(os.path.abspath('../../__file__'))
sys.path.insert(0, os.path.abspath(notebook_dir))

# Importar funciones del dataset y utilidades desde el paquete Utils
from Utils.ComparationGraphs import (
    compare_runs, 
    load_from_paths, 
    load_training_folder, 
    runs_metadata,
    compare_speedups,
    speedups_to_dataframe
)
```

##### Define functions for comparassion

```python
def main(folder: str = "../stats/ImagetNet"):

    runs = load_training_folder(folder)  

    print(runs_metadata(runs).to_string())

    chart = compare_runs(runs, loose=True)
    chart.display()
    return 0

def speedup_analysis(folder: str = "../stats/ImagetNet", base_case: str = None):

    runs = load_training_folder(folder)
    
    # Display metadata
    print(f"Speedup comparison (baseline: '{base_case}'):")
    print(runs_metadata(runs).to_string())
    
    # Create speedup visualization
    chart = compare_speedups(runs, base_case=base_case)
    chart.display()
    
    return runs
```
